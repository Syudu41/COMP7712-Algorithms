**Question 1 (10 points)**

Show that $5n^3 - 40n^2 \in \Theta(n^3)$. Find the specific upper bound and lower bound constants to prove this statement.  Show all of your steps clearly.


**Answer:**

To prove $f(n) = 5n^3 - 40n^2 \in \Theta(n^3)$, we need constants $c_1, c_2 > 0$ and $n_0 \geq 1$ such that:
$$c_1 \cdot n^3 \leq 5n^3 - 40n^2 \leq c_2 \cdot n^3 \quad \text{for all } n \geq n_0$$

(a) Upper bound: $5n^3 - 40n^2 \in O(n^3)$

For all $n \geq 1$, since $40n^2 \geq 0$:
$$5n^3 - 40n^2 \leq 5n^3$$

So **$c_2 = 5$, $n_0 = 1$**

(b) Lower Bound:** $5n^3 - 40n^2 \in \Omega(n^3)$

$c_1 \cdot n^3 \leq 5n^3 - 40n^2$, i.e., $(5 - c_1)n^3 \geq 40n^2$, which gives $n \geq \frac{40}{5 - c_1}$.

Pick $c_1 = 1$: $n \geq \frac{40}{4} = 10$.

So **$c_1 = 1$, $n_0 = 10$** 

(c) **Conclusion:** With $c_1 = 1$, $c_2 = 5$, $n_0 = 10$:
$$n^3 \leq 5n^3 - 40n^2 \leq 5n^3 \quad \forall n \geq 10$$

Therefore, $5n^3 - 40n^2 \in \Theta(n^3)$.

You are an investor with a fixed amount of capital $C$. There are $n$ startups looking for funding. Each startup requires a specific investment amount and promises a specific total profit. Since you are a solo investor, you can only invest in one startup at a time, and you cannot fund a project partially i.e. it is all or nothing.

**Key Constraint:** You can only invest once in each startup (either fully fund them or don't invest at all).

**Problem Setup:** Each startup investment opportunity is represented as a tuple: `(investment_needed, expected_annual_return)`
and you are have capital `C`.

**Example:**
Consider these startups:
```
Startup A: investment = $30,000, return = $12,000
Startup B: investment = $50,000, return = $20,000
Startup C: investment = $40,000, return = $15,000
```

**Strategy-1 (Invest in startup with the highest return)**

**Strategy-2 (Invest in startup with the smallest investment needed)**

**Part-A (15 points)**

Implement Greedy Strategy 1 to maximize the total profit/returns. In other words, choose the startup that offers the highest returns among those you can still afford to invest with your remaining capital.

````python
# Answer:
startups = [
    (50000, 15000),   # Startup A: needs $50k, returns $15k/year
    (30000, 10000),   # Startup B: needs $30k, returns $10k/year  
    (80000, 25000),   # Startup C: needs $80k, returns $25k/year
    (40000, 12000),   # Startup D: needs $40k, returns $12k/year
]
capital = 100000  # You have $100k to invest

# Returns total profit from the capital
def invest_greedy(startups, capital):
    # Sort in descending order to look at highest-return first 
    sorted_startups = sorted(startups, key=lambda s: s[1], reverse=True)
    
    total_profit = 0
    remaining = capital
    
    for investment, ret in sorted_startups:
        if investment <= remaining:
            total_profit += ret
            remaining -= investment
    
    return total_profit


print(invest_greedy(startups, capital))
````

**Part-C (15 points)**

**Implement an exhaustive strategy** that considers all possibilities.

**Strategy:** For each startup, try TWO possibilities:
1. Invest in it (if we can afford it)
2. Don't invest in it (skip it)

Then compare which choice leads to better overall returns.

**Implementation requirements:**
- Use recursion
- For the first startup in the list:
  - **If affordable**: Consider BOTH choices:
    - Choice 1: Invest in it -> add its return + recursively solve for remaining startups with reduced capital
    - Choice 2: Skip it -> recursively solve for remaining startups with same capital
    - Return the maximum of these two choices
  - **If NOT affordable**: Skip it -> recursively solve for remaining startups with same capital
- Base case: when startups list is empty, return 0

**Hints:**
- Use `startups[0]` to get the first startup
- Use `startups[1:]` to get the remaining startups
- Use `max()` to compare the two choices
- Think carefully about what you return in each case

````python
# Answer:
# Returns total profit from the capital
def invest_exhaustive(startups, capital):
    # Base case: no startups left to consider
    if len(startups) == 0:
        return 0
    
    investment, ret = startups[0]
    rest = startups[1:]
    
    if investment <= capital:
        # Choice 1: invest in it
        take = ret + invest_exhaustive(rest, capital - investment)
        # Choice 2: skip it
        skip = invest_exhaustive(rest, capital)
        return max(take, skip)
    else:
        # Cannot afford - skip it
        return invest_exhaustive(rest, capital)


startups = [
    (50000, 15000),  
    (30000, 10000),  
    (80000, 25000),   
    (40000, 12000),   
]
capital = 100000  
print("Greedy:", invest_greedy(startups, capital))
print("Exhaustive:", invest_exhaustive(startups, capital))
````

**Part-D (10 points)**

Write a **memoized** version of your solution from **Problem 3 Part-C** that caches results to avoid redundant calculations.

**Important:** Your cache key needs to capture all the information that affects the result. Think about what key you would need for the cache.


````python
# Answer:

def cache_invest_exh(startups, capital):
    cache = {} # key is the tuple pair of startup index and the capital remaining left to spend. 
    
    def helper(i, cap):
        # Base case: no more startups to consider
        if i == len(startups):
            return 0
        
        # Check cache
        if (i, cap) in cache:
            return cache[(i, cap)]
        
        investment, ret = startups[i]
        
        if investment <= cap:
            # Choice 1: invest in it
            take = ret + helper(i + 1, cap - investment)
            # Choice 2: skip it
            skip = helper(i + 1, cap)
            result = max(take, skip)
        else:
            # Cannot afford - skip
            result = helper(i + 1, cap)
        
        cache[(i, cap)] = result
        return result
    
    return helper(0, capital)

startups = [
    (50000, 15000),
    (30000, 10000),
    (80000, 25000),
    (40000, 12000),
]
capital = 100000

print("Memoized:", cache_invest_exh(startups, capital))
````

````python
# Starter code for you to use

import networkx as nx

def get_connected_gnp(n, p):
    while True:
        G = nx.gnp_random_graph(n, p, seed=423)
        if nx.is_connected(G):
            return G

# Use this backtracking template. 
def backtrack(solution, i, possibilities, is_feasible):
    if i == len(solution):
        print(solution)
    else:
        for possibility in possibilities:
            solution[i] = possibility
            if is_feasible(solution, i):
                backtrack(solution, i+1, possibilities, is_feasible)

# Create a fixed 6-node graph
# A higher p (0.7) ensures connectivity is easier to achieve
my_network = get_connected_gnp(6, 0.7)
nx.draw(my_network, with_labels=True)

# Answer:
# Define solution, possibilities, is_feasible and call the backtrack function.
solution = [None] * len(my_network.nodes)

possibilities = ['Red', 'Blue', 'Green']

def is_feasible(sol, i):
    for neighbor in my_network.neighbors(i):
        # Check node i's color against neighbors that are already colored
        if neighbor < i and sol[neighbor] == sol[i]: 
            return False
            
    return True

# call backtrack here:
print(backtrack(solution, 0, possibilities, is_feasible))
````
